```python
# Import Laibraries

import warnings
warnings.filterwarnings('ignore')

import re # Regular Expression

import pandas as pd # Dataframe
```

# Load Data :

```python
with open('chat.txt','r',encoding='utf-8') as f:
    data = f.read()
```

```python
# Regular Expression(regex)
pattern = r'\[\d{1,2}/\d{1,2}/\d{2,4},\s\d{1,2}:\d{2}:\d{2}\s[APap][Mm]\]\s'
```

```python
messages = re.split(pattern, data)[1:]
```

```python
dates = re.findall(pattern, data)
```

```python
df = pd.DataFrame({'user_message':messages, 'message_date':dates})
```

```python
df['message_date'] = (df['message_date']
                      .str.strip('[] ')
                      .str.replace('\u202f', ' ', regex=False))  # iOS sometimes uses a special space before AM/PM
df['message_date'] = pd.to_datetime(df['message_date'], format='%m/%d/%y, %I:%M:%S %p')

df.rename(columns={'message_date': 'date'}, inplace=True)
df.head()
```

# Data Pre-Processing :

```python
users, msgs = [], []
for m in df['user_message']:
    parts = re.split(r'^([^:]+):\s', m, maxsplit=1)
    if len(parts) > 1:
        users.append(parts[1]); msgs.append(parts[2])
    else:
        users.append('group_notification'); msgs.append(parts[0])
        
df['user'] = users
df['message'] = msgs
df.drop(columns=['user_message'], inplace=True)

df.head(3)
```

```python
df['year'] = df['date'].dt.year
```

```python
df['month'] = df['date'].dt.month_name()
```

```python
df['day'] = df['date'].dt.day
```

```python
df['hour'] = df['date'].dt.hour
```

```python
df['minunte'] = df['date'].dt.minute
```
